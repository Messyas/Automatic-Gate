```python
data_yaml = f"{dataset.location}/data.yaml"
print("Usando data config em:", data_yaml)
```

```python
model = YOLO("yolov8n.pt")
```

```python
model.train(
    data=data_yaml,      # arquivo .yaml com train/val paths e classes
    imgsz=640,           # resolução de entrada
    batch=16,            # tamanho de batch (ajuste à sua GPU)
    epochs=50,           # número de épocas
    lr0=0.01,            # learning rate inicial
    device=0,            # GPU 0 (em geral, device=0 usa primeira GPU CUDA)
    name="yolov8n_lp",   # nome da pasta onde logs/checkpoints serão salvos em runs/train/
    project="runs",      # pasta raiz para salvar
    augment=True         # ativa técnicas de data augmentation
)
```

```python
metrics = model.val()
print(metrics)
```
